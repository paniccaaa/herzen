```python
import pandas as pd

# Загрузка данных
df = pd.read_csv("train.csv")
```

```python
# 1. Количество мужчин и женщин на борту
male_count = df[df['Sex'] == 'male'].shape[0]
female_count = df[df['Sex'] == 'female'].shape[0]
print(male_count, female_count)
```

```python
# 2. Количество пассажиров, севших в разных портах
embarked_counts = df['Embarked'].value_counts()
s_count = embarked_counts.get('S', 0)
c_count = embarked_counts.get('C', 0)
q_count = embarked_counts.get('Q', 0)
print(s_count, c_count, q_count)
```

```python
# 3. Доля погибших пассажиров
total_passengers = df.shape[0]
survived_count = df['Survived'].sum()
died_count = total_passengers - survived_count
died_percentage = (died_count / total_passengers) * 100
print(died_count, round(died_percentage, 2))
```

```python
# 4. Доли пассажиров по классам
class_counts = df['Pclass'].value_counts(normalize=True) * 100
first_class = class_counts.get(1, 0)
second_class = class_counts.get(2, 0)
third_class = class_counts.get(3, 0)
print(round(first_class, 2), round(second_class, 2), round(third_class, 2))
```

```python
# 5. Коэффициент корреляции Пирсона между SibSp и Parch
sibsp_parch_corr = df[['SibSp', 'Parch']].corr().iloc[0, 1]
print(round(sibsp_parch_corr, 4))
```

```python
# 6. Корреляции с Survived
df['SexNumeric'] = df['Sex'].map({'male': 0, 'female': 1})  # Явное преобразование пола в числовой формат

age_survived_corr = df[['Age', 'Survived']].corr().iloc[0, 1]
sex_survived_corr = df[['SexNumeric', 'Survived']].corr().iloc[0, 1]  # Используем новую колонку
pclass_survived_corr = df[['Pclass', 'Survived']].corr().iloc[0, 1]

print(round(age_survived_corr, 4), round(sex_survived_corr, 4), round(pclass_survived_corr, 4))
```

```python
# 7. Возраст статистика
age_mean = df['Age'].mean()
age_median = df['Age'].median()
age_min = df['Age'].min()
age_max = df['Age'].max()
print(round(age_mean, 2), age_median, age_min, age_max)
```

```python
# 8. Цена билетов статистика
fare_mean = df['Fare'].mean()
fare_median = df['Fare'].median()
fare_min = df['Fare'].min()
fare_max = df['Fare'].max()
print(round(fare_mean, 2), round(fare_median, 2), fare_min, fare_max)
```

```python
# 9. Самое популярное мужское имя
df['CleanedName'] = df['Name'].str.extract(r',\s\w+\.\s(\w+)')
most_common_male_name = df[df['Sex'] == 'male']['CleanedName'].value_counts().idxmax()
print(most_common_male_name)
```

```python
import re

# Функция для корректного извлечения имен
def extract_name(name, sex):
    match = re.search(r"\(([^)]+)\)", name)  # Поиск имени в скобках
    if match and sex == 'female':  # Если это женщина, берем имя из скобок
        return match.group(1).split()[0]  # Берем только первое имя
    else:  # Иначе используем стандартное извлечение
        return re.search(r',\s\w+\.\s(\w+)', name).group(1) if re.search(r',\s\w+\.\s(\w+)', name) else None

# Добавляем корректные имена в новый столбец
df['CleanedName'] = df.apply(lambda row: extract_name(row['Name'], row['Sex']), axis=1)

# Находим самые популярные имена среди взрослых
adult_names = df[df['Age'] > 15]
male_adult_name = adult_names[adult_names['Sex'] == 'male']['CleanedName'].value_counts().idxmax()
female_adult_name = adult_names[adult_names['Sex'] == 'female']['CleanedName'].value_counts().idxmax()

print(male_adult_name, female_adult_name)
```
